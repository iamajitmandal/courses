#Copy vs View

```python
arr = np.array([1,2,3])
copy = arr.copy()
copy[0]=100
print(arr)
print(copy)
```

```python
view = arr.view()
view[0]=200
print(arr)
```

#Matrix Multilplication

```python
A=np.array([[1,2],
            [3,4]])

B=np.array([[5,6],
            [7,8]])

print(np.dot(A,B))
```

```python
print(A.T)
```

**Aggregate Functions**

```python
arr=np.array([1,2,3,4,5])

print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
print(np.std(arr))
print(np.var(arr))
```

```python
A=np.array([[1,2],[3,4]])
print(np.linalg.inv(A))
```

```python
print(np.linalg.det(A))
```

#MATPLOTLIB

```python
#Line Plot
import matplotlib.pyplot as plt
import numpy as np
x=np.array([1,2,3,4])
y=np.array([5,6,7,8])
plt.plot(x,y)
plt.show()

```

```python
#scatter plot
x=[1,2,3,4]
y=[2,5,4,7]
plt.scatter(x,y)
plt.show()
```

```python
#bar chart
courses=["Python","ML","DL"]
students=[40,30,20]
plt.bar(courses,students,width=0.5,color='black')
plt.show()
```

```python
#histogram-distribution of data
import matplotlib.pyplot as plt
marks = [45, 50, 55, 60, 60, 65, 70, 75, 80, 85]
plt.hist(marks)
plt.show()
```

```python
sizes = [100, 5, 30]
labels = ["Python", "ML", "DL"]
plt.pie(sizes, labels=labels)
plt.legend()
plt.show()
```

```python
x=np.array([1,2,3,4])
y=np.array([5,6,7,8])
plt.plot(x,y)
plt.title("Salary vs Experience")
plt.xlabel("Salary")
plt.ylabel("Experience")
```

```python
plt.figure(figsize=(2,2))
plt.plot([1,2,3],[2,4,6],color='g')
plt.show()
plt.savefig("graph.png")
```

```python
import matplotlib.pyplot as plt

ages = [18,20,22,23,21,25,24,19,20,21]

plt.hist(ages)

plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```
