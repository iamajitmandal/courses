```python
# Matplotlib: Matplotlib is the foundational library for creating static, interactive, and animated data visualizations in Python.
# It is widely used in data science, machine learning, and academic publishing because it offers total control over every element of a graph.
```

```python
# Matplotlib:
# Line Plot
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1, 2, 3, 4])
y = np.array([5, 6, 7, 8])
plt.plot(x, y)
plt.show()
```

```python
# Scatter Plot
# A scatter plot is a data visualization tool that uses dots to display values for two numeric variables.
# The independent variable is plotted on the x-axis, and the dependent variable is plotted on the y-axis.
# This arrangement instantly reveals relationships, correlations, and outliers between the two sets of data.
x = [1, 2, 3, 4]
y = [5, 6, 7, 8]
plt.scatter(x, y)
plt.show()
```

```python
# Bar Chart:
# A bar chart (or bar graph) is a data visualization that presents categorical data with rectangular bars.
# The length or height of each bar is proportional to the values they represent. They are primarily used to compare
# discrete categories at a glance.
courses = ['Python', 'ML', 'DL']
students = [40, 30, 20]
plt.bar(courses, students, width=0.5, color='red')
plt.show()
```

```python
# Histogram - distribution of data
# A histogram is a graphical display that shows the frequency distribution of continuous numerical data.
# Data is divided into continuous intervals called "bins" or "buckets" along the x-axis, while the y-axis represents the
# frequency or count of data points falling into each interval.
marks = [40, 50, 55, 60, 60, 65, 70, 75, 80, 85]
plt.hist(marks)
plt.show()
# In the histogram diagram, you can see that 2 students has secured the marks '60', 2 is shown on the left side
```

```python
# pie chart
# A pie chart is a circular statistical graphic that divides data into slices. Each slice’s size corresponds to its proportionate
# contribution to the total 100% (or 360°).
sizes = [40, 30, 30]
labels = ["Python", "ML", "DL"]
plt.pie(sizes, labels = labels)
plt.show()
```

```python
# You can customize the plots according to your design
# Line Plot
x = np.array([1, 2, 3, 4])
y = np.array([5, 6, 7, 8])
plt.plot(x, y)

# Customizing titles and labels
plt.title("Salary Vs Experience")
plt.xlabel("Salary")
plt.ylabel("Experience")

plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot([1,2,3],[2,4,6], color='r')
plt.show()
# To save as a figure
# plt.savefig("graph.png")
```

```python
# More Example: Generated Using AI
import matplotlib.pyplot as plt

# 1. Define your data
x_values = [1, 2, 3, 4, 5]
y_values = [2, 4, 6, 8, 10]

# 2. Create the line plot
plt.plot(x_values, y_values, marker='o', linestyle='-', color='b', label='Trend Line')

# 3. Customize titles and labels
plt.title("Simple Line Plot Example")
plt.xlabel("X-axis (Time/Steps)")
plt.ylabel("Y-axis (Values)")

# 4. Add a legend and grid
plt.legend()
plt.grid(True)

# 5. Display the plot
plt.show()
```
