```python
# Selecting rows and columns
# using iloc and loc
```

```python
import pandas as pd
data = {     "Name": ["Alice", "Bob", "Charlie", "David"],
              "Age": [23, 25, 22, 24],
             "City": ["NY", "LA", "Chicago", "Boston"] }
df = pd.DataFrame(data)
df
```

```python
# loc - label based indexing
# loc - full form is 'location'
# In the pandas library, loc stands for "location". It is a property used for label-based indexing to
# access specific rows and columns in a DataFrame or Series.
# select a row by a label
print(df.loc[0])
# print(df.loc[3])
```

```python
# Selecting multiple rows
print(df.loc[[0, 1]])
# print(df.loc[[0, 1, 2]])
```

```python
# select row and coloum
print(df.loc[2, "Name"])
```

```python
# Multiple columns
print(df.loc[:, ["Name", "Age"]])
```

```python
# slice rows
print(df.loc[1:3]) # gives row 0, 1, 2
```

```python
# using filter on Age
print(df.loc[df['Age'] > 23])
```

```python
# iloc - position base indexing
# iloc -> full form is 'Integer Location'
# In pandas, iloc is a property used for purely integer-based indexing. It allows you to select rows and
# columns from a DataFrame or Series by their numerical position (0-indexed), regardless of their text label
```

```python
# selecting first row
print(df.iloc[0])
```

```python
# row and column
print(df.iloc[2, 0])
```

```python
print(df.iloc[:, 0:2])
```

```python
# difference between loc and iloc
print("***** iloc *****\n")
print(df.loc[1:3])
print("\n")
print("***** iloc *****\n")
print(df.iloc[1:3])
```

```python
#   Name	    Age	  City
# 0	Alice	    23	  NY
# 1	Bob	      25	  LA
# 2	Charlie	  22	  Chicago
# 3	David	    24	  Boston

# [0, 3] -> 1st row and 4th row from above dataset
# [0, 2] -> 1st column and 3rd column from above dataset
print(df.iloc[[0, 3], [0, 2]])
```
