```python
import numpy as np
dic = {
    'Name': ['Bob','Sam','Mary','Alice','Beth'],
    'Status': ['Current','Alumni','Alumni','Alumni','Current'],
    'Fees': ['Due', 'Clear', 'Scholarship', 'Scholarship', 'Due'],
    'Score': [78,68,89,87,76],
    'Score2': [67,87,96,95,np.nan]
}
import pandas as pd
df = pd.DataFrame(dic)
df
```

```python
# grouping categorical data
status = df.groupby("Status")
status.groups
```

```python
fees = df.groupby("Fees")
# fees.groups
fees.groups.items() # gives in dictionary form
```

```python
for fee_status, students in fees.groups.items():
  # print(fee_status, students)
  if fee_status == "Scholarship":
    for student in students:
      #print(df.iloc[student])
      print(df.iloc[[student]])
      print("-" * 50)
      print("\n")
```

```python
df.groupby("Status").max()
```

```python
import numpy as np
dic = {
    'Name': ['Bob','Sam','Mary','Alice','Beth'],
    'Status': ['Current','Alumni','Alumni','Alumni','Current'],
    'Fees': ['Due', 'Clear', 'Scholarship', 'Scholarship', 'Due'],
    'Score': [78,68,89,87,76],
    'Score2': [67,87,96,95,np.nan],
    'Score3':[100,200,300,400,500]
}
import pandas as pd
df = pd.DataFrame(dic)
df
```

```python
status = df.groupby("Status")
```

```python
status[df.select_dtypes(include=np.number).columns].agg(['mean','median','std', 'min', 'max'])
```

```python
# merge: combine using common columns
import pandas as pd
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ram", "Hari", "Sita"]
})
marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [90, 95, 98]
})
result = pd.merge(students, marks, on="ID")
print(result)
```

```python
# inner join - default join and keeps only matching IDs
pd.merge(students, marks, on="ID", how="inner")
```

```python
# left join - keep all rows from left dataframe
pd.merge(students, marks, on="ID", how="left")
```

```python
# left join - keep all rows from right dataframe
pd.merge(students, marks, on="ID", how="right")
```

```python
# outer join
```

```python
# join
import pandas as pd
students = pd.DataFrame({
    "Name": ["Ram", "Hari", "Sita"]
}, index = [1, 2, 3])
marks = pd.DataFrame({
    "Marks": [90, 95, 98]
}, index = [1, 2, 3])
print(students.join(marks))
```

```python
# concatenation (using stack)
df1 = pd.DataFrame({
    "Name": ["Ram", "Hari"]
})
df2 = pd.DataFrame({
    "Name": ["Sita", "Gita"]
})
# pd.concat([df1, df2])
pd.concat([df1, df2], ignore_index=True)
```

```python
# concatenation using column wise
df1 = pd.DataFrame({
    "Name": ["Ram", "Hari"]
})
df2 = pd.DataFrame({
    "Name": ["Sita", "Gita"]
})
# pd.concat([df1, df2])
pd.concat([df1, df2], axis = 1)
```
