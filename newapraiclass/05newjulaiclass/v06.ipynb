```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
# Create data
x = np.array([[1], [2], [3], [4], [5]])
y = np.array([30, 40, 60, 95, 150])
```

```python
x
```

```python
y
```

```python
# create polynomial features
poly = PolynomialFeatures(degree=4) # use degree = 1,2 to see the difference (also run all to see the final output on different degrees)
x_poly = poly.fit_transform(x)
```

```python
x_poly
```

```python
# train the model
# create linear regression model
model = LinearRegression()
model.fit(x_poly, y)
```

```python
# predictin
predict = model.predict(x_poly)
```

```python
plt.scatter(x, y, color='red')
plt.plot(x, predict, color='blue')
plt.xlabel('experience')
plt.ylabel('salary')
plt.show()
```
