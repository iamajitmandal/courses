```python
# gradient calculation
x = torch.tensor(3.0, requires_grad=True)
y = x**2 # derivative of x**2 = 2x and 2x = 2*3 = 6
y.backward() # calculates how much y changes when x changes, and
# In PyTorch, y.backward() computes the gradients of a tensor with respect to the "leaf" nodes
# (the initial inputs or model parameters) in the dynamic computation graph. This function serves
# as the core engine for automatic differentiation and backpropagation during the training of neural networks.
print(x.grad)

# requires_grad = True
# In PyTorch, requires_grad = True tells the PyTorch Autograd Engine to track all mathematical operations
# performed on a tensor so it can automatically calculate gradients during backpropagation.
```
